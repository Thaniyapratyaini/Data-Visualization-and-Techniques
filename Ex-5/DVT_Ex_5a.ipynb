```python
# ============================================================
# EXPERIMENT 5a
# DATA VISUALIZATION USING MATPLOTLIB – SALES DATA
# ============================================================

import matplotlib.pyplot as plt

# ------------------------------------------------------------
# SALES DATA
# ------------------------------------------------------------

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 180, 170, 220, 250]
profit = [20, 30, 40, 35, 50, 60]

print("===== SALES DATA =====")
for i in range(len(months)):
    print(months[i], "Sales =", sales[i], "Profit =", profit[i])


# ============================================================
# PROGRAM 1: LINE CHART
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()


# ============================================================
# PROGRAM 2: BAR CHART
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()


# ============================================================
# PROGRAM 3: PIE CHART
# ============================================================

plt.figure(figsize=(8, 6))

plt.pie(
    sales,
    labels=months,
    autopct="%1.1f%%"
)

plt.title("Sales Distribution")

plt.show()


# ============================================================
# PROGRAM 4: HISTOGRAM
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(sales, bins=5)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()


# ============================================================
# PROGRAM 5: SCATTER PLOT
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()


# ============================================================
# PROGRAM 6: BOX PLOT
# ============================================================

plt.figure(figsize=(6, 5))

plt.boxplot(sales)

plt.title("Sales Box Plot")
plt.ylabel("Sales")

plt.show()


# ============================================================
# PROGRAM 7: MULTIPLE LINE CHARTS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")

plt.title("Sales and Profit")
plt.xlabel("Months")
plt.ylabel("Value")

plt.legend()

plt.show()


# ============================================================
# RESULT
# ============================================================

print("Data visualization using Matplotlib completed successfully.")
```
