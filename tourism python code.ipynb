# MINI PROJECT – TOURISM DATA ANALYTICS
Data loading, cleaning, analysis and visualization using Python.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Tourism_Data_Clean.csv")
tourism = df.copy()

print("First 5 Records:")
display(tourism.head())
print("\nDataset Shape:")
print(tourism.shape)
print("\nColumn Names:")
print(tourism.columns.tolist())
print("\nData Types:")
print(tourism.dtypes)
print("\nMissing Values:")
print(tourism.isnull().sum())
print("\nDuplicate Records:")
print(tourism.duplicated().sum())
```

```python
# Standardize column names and remove duplicates
tourism.columns = tourism.columns.str.lower().str.strip().str.replace(" ", "_")
before = tourism.shape[0]
tourism.drop_duplicates(inplace=True)
after = tourism.shape[0]

tourism["travel_date"] = pd.to_datetime(tourism["travel_date"])

print("Duplicates removed:", before-after)
print("\nCleaned Dataset Shape:", tourism.shape)
print("\nCleaned Column Names:", tourism.columns.tolist())
display(tourism.head())
```

```python
# Standardize categorical values
categorical_columns = [
    "destination","traveler_type","trip_type",
    "accommodation_type","transport_mode","booking_status"
]

for col in categorical_columns:
    tourism[col] = tourism[col].astype(str).str.strip().str.title()

tourism["year"] = tourism["travel_date"].dt.year
tourism["month"] = tourism["travel_date"].dt.month
tourism["month_name"] = tourism["travel_date"].dt.month_name()
tourism["year_month"] = tourism["travel_date"].dt.to_period("M").astype(str)

print("Destination values:")
print(tourism["destination"].value_counts())
print("\nBooking Status:")
print(tourism["booking_status"].value_counts())
```

```python
# Key tourism metrics
total_bookings = tourism["booking_id"].nunique()
total_revenue = tourism["amount"].sum()
average_trip_cost = tourism["amount"].mean()
total_tourists = tourism["customer_id"].nunique()
average_rating = tourism["rating"].mean()
completed_bookings = tourism.loc[tourism["booking_status"]=="Completed","booking_id"].nunique()

print("========== TOURISM ANALYTICS SUMMARY ==========")
print("Total Bookings :", total_bookings)
print("Total Revenue :", round(total_revenue,2))
print("Average Trip Cost :", round(average_trip_cost,2))
print("Total Tourists :", total_tourists)
print("Average Rating :", round(average_rating,2))
print("Completed Bookings :", completed_bookings)
```

```python
# Destination analysis
destination_analysis = tourism.groupby("destination").agg(
    Bookings=("booking_id","nunique"),
    Total_Revenue=("amount","sum"),
    Average_Cost=("amount","mean"),
    Average_Rating=("rating","mean")
).reset_index()

display(destination_analysis.sort_values("Total_Revenue", ascending=False))
```

```python
# Traveler type analysis
traveler_analysis = tourism.groupby("traveler_type").agg(
    Bookings=("booking_id","nunique"),
    Total_Revenue=("amount","sum"),
    Average_Cost=("amount","mean")
).reset_index()

display(traveler_analysis)
```

```python
# Accommodation and transport analysis
accommodation_analysis = tourism.groupby("accommodation_type").agg(
    Bookings=("booking_id","nunique"),
    Total_Revenue=("amount","sum")
).reset_index()

transport_analysis = tourism.groupby("transport_mode").agg(
    Bookings=("booking_id","nunique"),
    Total_Revenue=("amount","sum")
).reset_index()

display(accommodation_analysis)
display(transport_analysis)
```

```python
# Monthly tourism analysis
monthly_analysis = tourism.groupby("year_month").agg(
    Bookings=("booking_id","nunique"),
    Total_Revenue=("amount","sum"),
    Average_Cost=("amount","mean")
).reset_index()

display(monthly_analysis)
```

```python
# Visualizations
destination_revenue = tourism.groupby("destination")["amount"].sum().sort_values(ascending=False)
destination_revenue.plot(kind="bar")
plt.title("Revenue by Destination")
plt.xlabel("Destination")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
booking_type = tourism.groupby("traveler_type")["booking_id"].nunique().sort_values(ascending=False)
booking_type.plot(kind="bar")
plt.title("Bookings by Traveler Type")
plt.xlabel("Traveler Type")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
transport_bookings = tourism.groupby("transport_mode")["booking_id"].nunique()
transport_bookings.plot(kind="pie", autopct="%1.1f%%")
plt.title("Bookings by Transport Mode")
plt.ylabel("")
plt.show()
```

```python
monthly_revenue = tourism.groupby("year_month")["amount"].sum()
monthly_revenue.plot(kind="line", marker="o")
plt.title("Monthly Tourism Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
status_bookings = tourism.groupby("booking_status")["booking_id"].nunique()
status_bookings.plot(kind="bar")
plt.title("Booking Status Analysis")
plt.xlabel("Booking Status")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Export processed dataset for Power BI and Tableau
tourism.to_csv("Tourism_Data_Clean_Processed.csv", index=False)
print("Processed tourism dataset exported successfully.")
```
