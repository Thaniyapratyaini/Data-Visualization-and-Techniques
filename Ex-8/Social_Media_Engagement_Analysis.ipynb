## Aim:
To clean, transform, and analyze social media engagement data using Python, Power BI, and Tableau to derive meaningful insights into platform trends, content types, engagement metrics, sentiment, and user activity.


## Algorithm:
1. Load the social media engagement dataset from the CSV file using Pandas.
2. Inspect the dataset for its structure, data types, missing values, and duplicate records.
3. Clean the dataset by removing duplicate records and standardizing column names.
4. Convert the post date into a proper date format.
5. Standardize categorical values such as platform, content type, category, and sentiment.
6. Create derived fields such as year, month, month name, and year-month for time-based analysis.
7. Analyze key social media metrics such as total posts, total engagements, average engagement rate, and total users.

## Architecture:
CSV Dataset → Python Data Loading & Cleaning → Engagement Analysis → Power BI Dashboard / Tableau Dashboard → Insights


## Program:

### 1. Python – Data Loading, Inspection and Cleaning


```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("/content/Social_Media_Engagement_Clean.csv")

# Create a copy for processing
social = df.copy()

print("First 5 Records:")
display(social.head())

print("\nDataset Shape:")
print(social.shape)

print("\nColumn Names:")
print(social.columns.tolist())

print("\nData Types:")
print(social.dtypes)

print("\nMissing Values:")
print(social.isnull().sum())

print("\nDuplicate Records:")
print(social.duplicated().sum())

# Create another copy for cleaning
engagement = social.copy()

# Standardize column names
engagement.columns = (
    engagement.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

# Remove duplicate records
before = engagement.shape[0]
engagement.drop_duplicates(inplace=True)
after = engagement.shape[0]

print("Duplicates removed:", before - after)

print("\nMissing Values After Cleaning:")
print(engagement.isnull().sum())

print("\nCleaned Dataset Shape:")
print(engagement.shape)

print("\nCleaned Column Names:")
print(engagement.columns.tolist())

print("\nFirst 5 Cleaned Records:")
display(engagement.head())
```

### 2. Python – Social Media Engagement Analysis


```python
# Total number of posts
total_posts = engagement["post_id"].nunique()

# Total engagements
total_engagements = (
    engagement["likes"] +
    engagement["comments"] +
    engagement["shares"]
).sum()

# Average engagement rate
average_engagement_rate = engagement["engagement_rate"].mean()

# Total users
total_users = engagement["user_id"].nunique()

# Total likes
total_likes = engagement["likes"].sum()

# Total comments
total_comments = engagement["comments"].sum()

# Total shares
total_shares = engagement["shares"].sum()

print("========== SOCIAL MEDIA ANALYTICS SUMMARY ==========")
print("Total Posts :", total_posts)
print("Total Engagements :", total_engagements)
print("Average Engagement Rate :", round(average_engagement_rate, 2), "%")
print("Total Users :", total_users)
print("Total Likes :", total_likes)
print("Total Comments :", total_comments)
print("Total Shares :", total_shares)
```

### 3. Python – Standardization of Categorical Values


```python
categorical_columns = [
    "platform",
    "content_type",
    "category",
    "sentiment"
]

for col in categorical_columns:
    engagement[col] = (
        engagement[col]
        .astype(str)
        .str.strip()
        .str.lower()
        .str.title()
    )

print("Platforms:")
print(engagement["platform"].value_counts())

print("\nContent Types:")
print(engagement["content_type"].value_counts())

print("\nSentiment:")
print(engagement["sentiment"].value_counts())
```

### 4. Python – Platform Analysis


```python
platform_analysis = engagement.groupby("platform").agg(
    Posts=("post_id", "nunique"),
    Total_Engagements=("likes", "sum"),
    Average_Engagement_Rate=("engagement_rate", "mean")
).reset_index()

# Add comments and shares to total engagement
extra = engagement.groupby("platform")[["comments", "shares"]].sum().reset_index()
platform_analysis["Total_Engagements"] = (
    platform_analysis["Total_Engagements"]
    + extra["comments"]
    + extra["shares"]
)

print("Platform Analysis:")
display(platform_analysis)
```

### 5. Python – Content Type Analysis


```python
content_analysis = engagement.groupby("content_type").agg(
    Posts=("post_id", "nunique"),
    Likes=("likes", "sum"),
    Comments=("comments", "sum"),
    Shares=("shares", "sum"),
    Average_Engagement_Rate=("engagement_rate", "mean")
).reset_index()

print("Content Type Analysis:")
display(content_analysis)
```

### 6. Python – Monthly Engagement Analysis


```python
engagement["post_date"] = pd.to_datetime(engagement["post_date"])

engagement["year_month"] = (
    engagement["post_date"]
    .dt.to_period("M")
    .astype(str)
)

monthly_analysis = engagement.groupby("year_month").agg(
    Posts=("post_id", "nunique"),
    Likes=("likes", "sum"),
    Comments=("comments", "sum"),
    Shares=("shares", "sum"),
    Average_Engagement_Rate=("engagement_rate", "mean")
).reset_index()

monthly_analysis["Total_Engagements"] = (
    monthly_analysis["Likes"] +
    monthly_analysis["Comments"] +
    monthly_analysis["Shares"]
)

print("Monthly Engagement Analysis:")
display(monthly_analysis)
```

### 7. Python – Data Visualization

#### Platform-wise Engagement


```python
platform_engagement = engagement.groupby("platform").agg(
    Likes=("likes", "sum"),
    Comments=("comments", "sum"),
    Shares=("shares", "sum")
)

platform_engagement.plot(kind="bar")
plt.title("Platform-wise Social Media Engagement")
plt.xlabel("Platform")
plt.ylabel("Engagement Count")
plt.xticks(rotation=0)
plt.show()
```

#### Engagement by Content Type


```python
content_engagement = engagement.groupby("content_type")[
    ["likes", "comments", "shares"]
].sum()

content_engagement.plot(kind="bar")
plt.title("Engagement by Content Type")
plt.xlabel("Content Type")
plt.ylabel("Engagement Count")
plt.xticks(rotation=0)
plt.show()
```

#### Engagement by Platform


```python
platform_amount = engagement.groupby("platform")[
    ["likes", "comments", "shares"]
].sum().sum(axis=1)

platform_amount.plot(kind="pie", autopct="%1.1f%%")
plt.title("Engagement Contribution by Platform")
plt.ylabel("")
plt.show()
```

#### Monthly Engagement Trend


```python
monthly_engagement = engagement.groupby("year_month")[
    ["likes", "comments", "shares"]
].sum().sum(axis=1)

monthly_engagement.plot(kind="line", marker="o")
plt.title("Monthly Social Media Engagement Trend")
plt.xlabel("Month")
plt.ylabel("Total Engagement")
plt.xticks(rotation=45)
plt.show()
```
