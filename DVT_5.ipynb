```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("retail_sales_dataset.csv")


# ============================================================
# EXPERIMENT 5
# EDA - DATA INSPECTION AND ANALYSIS
# ============================================================


# ============================================================
# 1. VIEWING AND INSPECTING DATAFRAME
# ============================================================

print("========== FIRST 5 ROWS ==========")
print(df.head())

print("\n========== FIRST 10 ROWS ==========")
print(df.head(10))

print("\n========== LAST 5 ROWS ==========")
print(df.tail())

print("\n========== DATASET SHAPE ==========")
print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])

print("\n========== COLUMN NAMES ==========")
print(df.columns.tolist())

print("\n========== DATA TYPES ==========")
print(df.dtypes)

print("\n========== DATAFRAME INFORMATION ==========")
df.info()


# ============================================================
# 2. DATA INSPECTION
# ============================================================

print("\n========== STATISTICAL SUMMARY ==========")
print(df.describe())

print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

print("\n========== DUPLICATE ROWS ==========")
print("Number of duplicate rows:", df.duplicated().sum())

print("\n========== UNIQUE VALUES ==========")
print(df.nunique())


# ============================================================
# 3. FILTERING DATA USING CONDITIONS
# ============================================================

# Customers above age 40

age_above_40 = df[df["age"] > 40]

print("\n========== CUSTOMERS ABOVE AGE 40 ==========")
print(age_above_40.head())


# Orders with quantity greater than 5

high_quantity = df[df["quantity"] > 5]

print("\n========== QUANTITY GREATER THAN 5 ==========")
print(high_quantity.head())


# Orders with sales amount greater than 1000

high_sales = df[df["sales_amount"] > 1000]

print("\n========== SALES AMOUNT GREATER THAN 1000 ==========")
print(high_sales.head())


# Profitable transactions

profitable = df[df["profit"] > 0]

print("\n========== PROFITABLE TRANSACTIONS ==========")
print(profitable.head())


# Discount greater than 10%

high_discount = df[df["discount_pct"] > 10]

print("\n========== DISCOUNT GREATER THAN 10% ==========")
print(high_discount.head())


# ============================================================
# 4. FILTERING USING MULTIPLE CONDITIONS
# ============================================================

# Age greater than 40 AND sales greater than 1000

condition1 = df[
    (df["age"] > 40) &
    (df["sales_amount"] > 1000)
]

print("\n========== AGE > 40 AND SALES > 1000 ==========")
print(condition1.head())


# Quantity greater than 5 AND profit greater than 100

condition2 = df[
    (df["quantity"] > 5) &
    (df["profit"] > 100)
]

print("\n========== QUANTITY > 5 AND PROFIT > 100 ==========")
print(condition2.head())


# ============================================================
# 5. SUBSETTING DATA
# ============================================================

# Select specific columns

subset1 = df[
    [
        "customer_name",
        "product_name",
        "quantity",
        "sales_amount"
    ]
]

print("\n========== SELECTED COLUMNS ==========")
print(subset1.head())


# Customer and sales information

subset2 = df[
    [
        "customer_id",
        "customer_name",
        "sales_amount",
        "profit"
    ]
]

print("\n========== CUSTOMER SALES INFORMATION ==========")
print(subset2.head())


# ============================================================
# 6. MEASURES OF CENTRAL TENDENCY
# ============================================================

# -------------------------
# MEAN
# -------------------------

print("\n========== MEAN ==========")

print("Mean Age          :", df["age"].mean())
print("Mean Quantity     :", df["quantity"].mean())
print("Mean Unit Price   :", df["unit_price"].mean())
print("Mean Sales Amount :", df["sales_amount"].mean())
print("Mean Profit       :", df["profit"].mean())


# -------------------------
# MEDIAN
# -------------------------

print("\n========== MEDIAN ==========")

print("Median Age          :", df["age"].median())
print("Median Quantity     :", df["quantity"].median())
print("Median Unit Price   :", df["unit_price"].median())
print("Median Sales Amount :", df["sales_amount"].median())
print("Median Profit       :", df["profit"].median())


# -------------------------
# MODE
# -------------------------

print("\n========== MODE ==========")

print("Mode Age:")
print(df["age"].mode().tolist())

print("\nMode Quantity:")
print(df["quantity"].mode().tolist())

print("\nMode Unit Price:")
print(df["unit_price"].mode().tolist())

print("\nMode Sales Amount:")
print(df["sales_amount"].mode().tolist())

print("\nMode Profit:")
print(df["profit"].mode().tolist())


# ============================================================
# 7. MEASURES OF DISPERSION
# ============================================================

# -------------------------
# RANGE
# -------------------------

print("\n========== RANGE ==========")

age_range = df["age"].max() - df["age"].min()

quantity_range = (
    df["quantity"].max() -
    df["quantity"].min()
)

price_range = (
    df["unit_price"].max() -
    df["unit_price"].min()
)

sales_range = (
    df["sales_amount"].max() -
    df["sales_amount"].min()
)

profit_range = (
    df["profit"].max() -
    df["profit"].min()
)

print("Age Range          :", age_range)
print("Quantity Range     :", quantity_range)
print("Unit Price Range   :", price_range)
print("Sales Amount Range :", sales_range)
print("Profit Range       :", profit_range)


# -------------------------
# VARIANCE
# -------------------------

print("\n========== VARIANCE ==========")

print(
    "Age Variance          :",
    df["age"].var()
)

print(
    "Quantity Variance     :",
    df["quantity"].var()
)

print(
    "Unit Price Variance   :",
    df["unit_price"].var()
)

print(
    "Sales Amount Variance :",
    df["sales_amount"].var()
)

print(
    "Profit Variance       :",
    df["profit"].var()
)


# -------------------------
# STANDARD DEVIATION
# -------------------------

print("\n========== STANDARD DEVIATION ==========")

print(
    "Age Standard Deviation          :",
    df["age"].std()
)

print(
    "Quantity Standard Deviation     :",
    df["quantity"].std()
)

print(
    "Unit Price Standard Deviation   :",
    df["unit_price"].std()
)

print(
    "Sales Amount Standard Deviation :",
    df["sales_amount"].std()
)

print(
    "Profit Standard Deviation       :",
    df["profit"].std()
)


# ============================================================
# 8. COMPLETE DESCRIPTIVE STATISTICS
# ============================================================

print("\n========== COMPLETE DESCRIPTIVE STATISTICS ==========")

numeric_columns = [
    "age",
    "quantity",
    "unit_price",
    "discount_pct",
    "sales_amount",
    "profit",
    "shipping_cost",
    "days_to_ship"
]

descriptive_stats = df[numeric_columns].describe().T

descriptive_stats["range"] = (
    descriptive_stats["max"] -
    descriptive_stats["min"]
)

descriptive_stats["variance"] = (
    df[numeric_columns].var()
)

descriptive_stats["mode"] = [
    df[column].mode().iloc[0]
    for column in numeric_columns
]

print(descriptive_stats)


# ============================================================
# 9. VISUALIZATION
# ============================================================

# Sales Amount Distribution

plt.figure(figsize=(8, 5))

sns.histplot(
    df["sales_amount"],
    kde=True
)

plt.title("Sales Amount Distribution")

plt.xlabel("Sales Amount")

plt.ylabel("Frequency")

plt.show()


# Sales Amount Boxplot

plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df["sales_amount"]
)

plt.title("Sales Amount Boxplot")

plt.xlabel("Sales Amount")

plt.show()


# Profit Distribution

plt.figure(figsize=(8, 5))

sns.histplot(
    df["profit"],
    kde=True
)

plt.title("Profit Distribution")

plt.xlabel("Profit")

plt.ylabel("Frequency")

plt.show()


# ============================================================
# 10. FINAL SUMMARY
# ============================================================

print("\n================================================")
print("EXPERIMENT 5 COMPLETED SUCCESSFULLY!")
print("================================================")

print("\nTopics Covered:")

print("1. Data Inspection and Analysis")
print("2. Viewing and Inspecting DataFrames")
print("3. Filtering Data using Conditions")
print("4. Subsetting Data")
print("5. Mean")
print("6. Median")
print("7. Mode")
print("8. Range")
print("9. Variance")
print("10. Standard Deviation")
print("11. Descriptive Statistics")
print("12. Data Visualization")
```
