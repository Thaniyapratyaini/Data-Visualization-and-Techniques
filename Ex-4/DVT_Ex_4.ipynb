```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("retail_sales_dataset.csv")
```

```python


print("First 5 Rows")
print(df.head())

print("\nShape")
print(df.shape)

print("\nData Types")
print(df.dtypes)

# --------------------------------------------------
# Statistical Summary
# --------------------------------------------------

print("\nStatistical Summary")

print(df.describe())

# --------------------------------------------------
# High Sales Transactions
# --------------------------------------------------

high_sales = df[df["sales_amount"] > 1000]

print("\nHigh Sales Transactions")

print(high_sales.head())

# --------------------------------------------------
# Orders from South Region
# --------------------------------------------------

south_region = df[df["region"] == "South"]

print("\nSouth Region Orders")

print(south_region.head())

# --------------------------------------------------
# Sort by Sales Amount
# --------------------------------------------------

sorted_sales = df.sort_values(
    by="sales_amount",
    ascending=False
)

print("\nTop Sales")

print(sorted_sales.head())

# --------------------------------------------------
# Basic Statistics
# --------------------------------------------------

print("\nSales Statistics")

print("Mean :", df["sales_amount"].mean())

print("Median :", df["sales_amount"].median())

print("Mode :")

print(df["sales_amount"].mode())

print("Standard Deviation :", df["sales_amount"].std())
```

```python

# --------------------------------------------------
# Average Sales by Region
# --------------------------------------------------

region_avg = df.groupby("region")["sales_amount"].mean()

print("\nAverage Sales by Region")

print(region_avg.sort_values(ascending=False))

# --------------------------------------------------
# Bar Chart
# --------------------------------------------------

plt.figure(figsize=(8,5))

region_avg.sort_values(
    ascending=False
).plot(kind="bar")

plt.title("Average Sales by Region")

plt.ylabel("Average Sales")

plt.xlabel("Region")

plt.show()

# --------------------------------------------------
# Correlation Matrix
# --------------------------------------------------

corr = df[
    [
        "age",
        "quantity",
        "unit_price",
        "discount_pct",
        "sales_amount",
        "profit",
        "shipping_cost",
        "days_to_ship"
    ]
].corr()

print("\nCorrelation Matrix")

print(corr)

# --------------------------------------------------
# Heatmap
# --------------------------------------------------

plt.figure(figsize=(9,7))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()

# --------------------------------------------------
# Scatter Plot
# --------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="unit_price",
    y="sales_amount"
)

plt.title("Unit Price vs Sales Amount")

plt.xlabel("Unit Price")

plt.ylabel("Sales Amount")

plt.show()
```
