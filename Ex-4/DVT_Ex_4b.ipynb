```python
import pandas as pd
import matplotlib.pyplot as plt

# Import Dataset
df = pd.read_csv("/content/city_day.csv")

# Dataset Inspection
print("===== FIRST FIVE RECORDS =====")
display(df.head())

print("\n===== SHAPE OF DATASET =====")
print(df.shape)

print("\n===== DATA TYPES =====")
print(df.dtypes)

print("\n===== SUMMARY =====")
display(df.describe())

# Missing Values
print("\n===== MISSING VALUES =====")
print(df.isnull().sum())

# Filtering Data

# Records with PM2.5 greater than 100
print("\n===== RECORDS WITH PM2.5 > 100 =====")
high_pm25 = df[df["PM2.5"] > 100]
display(high_pm25.head())

# Records with PM10 greater than 100
print("\n===== RECORDS WITH PM10 > 100 =====")
high_pm10 = df[df["PM10"] > 100]
display(high_pm10.head())

# Sorting Data
print("\n===== SORT BY PM2.5 (DESCENDING) =====")
sorted_df = df.sort_values(by="PM2.5", ascending=False)
display(sorted_df.head())

# Statistical Analysis
print("\n===== PM2.5 STATISTICS =====")

print("Mean :", df["PM2.5"].mean())

print("Median :", df["PM2.5"].median())

print("Mode :")
print(df["PM2.5"].mode())

print("Standard Deviation :", df["PM2.5"].std())

print("\n===== DESCRIPTIVE STATISTICS =====")
display(df.describe())

# Grouping and Aggregation
print("\n===== AVERAGE PM2.5 BY CITY =====")
city_avg_pm25 = df.groupby("City")["PM2.5"].mean()
display(city_avg_pm25)

print("\n===== AVERAGE PM10 BY CITY =====")
city_avg_pm10 = df.groupby("City")["PM10"].mean()
display(city_avg_pm10)

# Correlation Analysis
print("\n===== CORRELATION =====")

corr = df[
    ["PM2.5", "PM10", "NO", "NO2", "NOx", "CO", "SO2", "O3",
     "Benzene", "Toluene", "Xylene"]
].corr()

display(corr)

# Visualization 1
city_avg_pm25.plot(kind="bar", figsize=(10,5))

plt.title("Average PM2.5 by City")
plt.xlabel("City")
plt.ylabel("Average PM2.5")
plt.xticks(rotation=45)
plt.show()

# Visualization 2
plt.figure(figsize=(8,5))

plt.scatter(df["PM2.5"], df["PM10"])

plt.title("PM2.5 vs PM10")
plt.xlabel("PM2.5")
plt.ylabel("PM10")

plt.show()
```
